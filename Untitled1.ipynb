```python
pair_counts = df['페어'].value_counts().reset_index()
pair_counts.columns = ['페어', '출연횟수']
pair_counts['비중(%)'] = (pair_counts['출연횟수'] / pair_counts['출연횟수'].sum() * 100).round(1)
pair_counts
```

```python
plt.figure(figsize=(9, 5))
bars = plt.bar(pair_counts['페어'], pair_counts['출연횟수'], color='#4472C4')
plt.title('뮤지컬 <베토벤> 페어별 출연 횟수 (2026.06.09~08.11)', fontsize=14, fontweight='bold')
plt.xlabel('페어 (루드비히-토니)')
plt.ylabel('출연 횟수')
plt.xticks(rotation=20)

for b in bars:
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h + 0.3, f'{int(h)}회', ha='center', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
df['월'] = df['날짜'].dt.strftime('%Y-%m')
monthly = df.groupby(['월', '페어']).size().unstack(fill_value=0)

plt.figure(figsize=(9, 5))
for pair in monthly.columns:
    plt.plot(monthly.index, monthly[pair], marker='o', label=pair)

plt.title('월별 페어 출연 추이', fontsize=14, fontweight='bold')
plt.xlabel('월')
plt.ylabel('출연 횟수')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=9)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()

monthly
```

```python
import seaborn as sns

# 요일 x 페어 집계표 만들기
day_order = ['월', '화', '수', '목', '금', '토', '일']
pair_order = df['페어'].unique().tolist()

pivot = df.pivot_table(index='요일', columns='페어', values='날짜', aggfunc='count', fill_value=0)
pivot = pivot.reindex(index=[d for d in day_order if d in pivot.index])

plt.figure(figsize=(9, 5))
sns.heatmap(pivot, annot=True, fmt='d', cmap='Blues', linewidths=0.5, linecolor='white',
            cbar_kws={'label': '출연 횟수'})
plt.title('요일 × 페어 히트맵', fontsize=14, fontweight='bold')
plt.xlabel('페어')
plt.ylabel('요일')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()

pivot
```
